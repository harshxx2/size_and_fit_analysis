# 02 Data Cleaning

This notebook applies the cleaning pipeline defined in `src/data_cleaning.py` to the raw dataset and writes the processed CSV. It also creates derived columns used for analysis.


```python
import pandas as pd
from pathlib import Path
import sys
sys.path.append('../src')
from data_cleaning import clean_data

raw_path = Path('../data/raw/renttherunway_final_data.json')
df_raw = pd.read_json(raw_path, lines=True)
df_clean = clean_data(df_raw)
out_path = Path('../data/processed/renttherunway_cleaned.csv')
df_clean.to_csv(out_path, index=False)
print('Cleaned data written to', out_path)
```

## Outcome
* Rows: **192,544** (unchanged)
* Added columns: `height_cm`, `weight_kg`, `bmi`, `size_group`, `fit_numeric`.
* Missing values reduced to < 2 % for numeric fields.
